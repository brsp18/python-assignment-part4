```python

!pip install scikit-learn

import pandas as pd
from sklearn.model_selection import train_test_split # 1. Import the libraries

df= pd.read_csv("students.csv") #2. Import the data set
feature_selection = ['math', 'science', 'english', 
                     'history', 'pe', 'attendance_pct', 
                     'study_hours_per_day'] # 3. Selecting the columns(features) from data set

x= df[feature_selection]  # 4. insert the selcted features to dataframe
y= df['passed']

print("\n")
print("Features Shape", x.shape)
print("\n")
print("Target Shape ", y.shape)
print("\n")
print(x.head())
print("\n")
print(y.head())







```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42) 
                   # testing & training the data at 80%(train)-20%(test)
print("X train shape", x_train.shape)
print("Y train shape", y_train.shape)
print("X test Shape", x_test.shape)
print("Y test shape", y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.fit_transform(x_test)

print("X train scaled Shape", x_train_scaled.shape)
print("X test scaled shape",x_test_scaled.shape)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(random_state=42)
model.fit(x_train_scaled, y_train)  # fit() used for Training the model

train_accuracy = model.score(x_train_scaled, y_train) #Print training accuracy
print("Training Accuracy:", train_accuracy * 100, "%")
```

```python
y_pred = model.predict(x_test_scaled) # predic() used on test data for Predicting 

test_accuracy = model.score(x_test_scaled, y_test) # Printing Test accuracy
print(f"Test Accuracy: {test_accuracy * 100:.2f}%")

print("\n===== Student Predictions =====") 
names = df.loc[x_test.index, 'name'] # getting syudent name

for name, actual, predicted in zip(names, y_test, y_pred):
    result = "Correct" if actual == predicted else "Wrong"
    print(f"{name:10} | Actual: {actual} | Predicted: {predicted} | {result}")

```

```python
new_student = [[75, 70, 68, 65, 80, 82, 3.2]] # defining new student
new_student_scaled = scaler.transform(new_student) # only transformin, not fitting

prediction = model.predict(new_student_scaled) #Predicting New student
probability = model.predict_proba(new_student_scaled)

result = "Pass" if prediction[0] == 1 else "Fail" #printing results#
print(f"Prediction        : {result}")
print(f"Probability Fail  : {probability[0][0] * 100:.2f}%")
print(f"Probability Pass  : {probability[0][1] * 100:.2f}%")


```
